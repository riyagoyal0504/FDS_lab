```python
import pandas as pd
import numpy as np
file_name = list(uploaded.keys())[0]

df = pd.read_csv("UP_Heatwave_RAW_data_clean (4).csv")

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)

df.head()
```

```python
print("Column names:")
print(df.columns)
```

```python
print("Null values in each column:")
print(df.isnull().sum())

print("\nTotal null values:", df.isnull().sum().sum())
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column].fillna(df[column].mean(), inplace=True)

print("Missing values handled successfully.")

print(df.isnull().sum())
```

```python
top_8_temperature = df.sort_values(
    by='T2M_MAX',
    ascending=False
).head(8)

display(top_8_temperature[['date', 'city', 'T2M_MAX']])
```

```python
hot_days = df[df['T2M_MAX'] > 40]

display(hot_days[['date', 'city', 'T2M_MAX']].head(8))
```

```python
city_frequency = df['city'].value_counts()

print("Frequency of each city:")
print(city_frequency)
```

```python
temperature_category = pd.cut(
    df['T2M_MAX'],
    bins=[-np.inf, 30, 35, 40, np.inf],
    labels=['Below 30', '30-35', '35-40', 'Above 40']
)

print("Frequency of temperature categories:")
print(temperature_category.value_counts())
```

```python
sorted_rows = df.sort_values(
    by=['city', 'T2M_MAX'],
    ascending=[True, False]
)

display(sorted_rows[['city', 'date', 'T2M_MAX']].head(10))
```

```python
sorted_columns = df.sort_index(axis=1)

display(sorted_columns.head())
```

```python
print("First row:")
display(df.iloc[0])

print("\nFirst five rows:")
display(df.iloc[0:5])
```

```python
print("First five rows using label-based indexing:")
display(df.loc[0:4, ['date', 'city', 'T2M_MAX']])
```

```python
case_1 = df[
    (df['T2M_MAX'] > 40) &
    (df['RH2M'] < 50)
]

display(case_1[['date', 'city', 'T2M_MAX', 'RH2M']].head(8))
```

```python
case_2 = df[
    (df['T2M_MAX'] > 35) &
    (df['WS10M'] > 5)
]

display(case_2[['date', 'city', 'T2M_MAX', 'WS10M']].head(8))
```

```python
case_3 = df[
    (df['T2M_MAX'] > 40) &
    (df['RH2M'] > 60) &
    (df['WS10M'] < 5)
]

display(case_3[['date', 'city', 'T2M_MAX', 'RH2M', 'WS10M']].head(8))
```

```python
print("Minimum maximum temperature:", df['T2M_MAX'].min())
print("Maximum maximum temperature:", df['T2M_MAX'].max())

print("\nMinimum minimum temperature:", df['T2M_MIN'].min())
print("Maximum minimum temperature:", df['T2M_MIN'].max())
```

```python
min_temperature = df.loc[df['T2M_MAX'].idxmin()]
max_temperature = df.loc[df['T2M_MAX'].idxmax()]

print("Record with minimum maximum temperature:")
display(min_temperature[['date', 'city', 'T2M_MAX']])

print("\nRecord with maximum maximum temperature:")
display(max_temperature[['date', 'city', 'T2M_MAX']])
```

```python
city_temperature = df.groupby('city')['T2M_MAX'].mean()

print("Average maximum temperature by city:")
print(city_temperature)
```

```python
df['year'] = pd.to_datetime(df['date']).dt.year

city_year_temperature = df.groupby(
    ['city', 'year']
)['T2M_MAX'].mean()

display(city_year_temperature.head(10))
```

```python
city_analysis = df.groupby('city').agg({
    'T2M_MAX': ['mean', 'min', 'max'],
    'RH2M': 'mean'
})

display(city_analysis)
```

```python
city_weather = df.groupby('city').agg({
    'T2M': 'mean',
    'WS10M': 'mean',
    'PRECTOTCORR': 'sum'
})

display(city_weather)
```

```python
selected_groups = df.groupby('city').filter(
    lambda x: x['T2M_MAX'].mean() > 35
)

display(
    selected_groups[['city', 'date', 'T2M_MAX']].head(10)
)
```

```python
df['Temperature_Range'] = df['T2M_MAX'] - df['T2M_MIN']

display(
    df[['date', 'city', 'T2M_MAX', 'T2M_MIN', 'Temperature_Range']].head()
)
```

```python
 correlation = df['T2M_MAX'].corr(df['RH2M'])

print(
    "Correlation between maximum temperature and relative humidity:",
    correlation
)
```

```python
temperature_array = np.array(
    df['T2M_MAX'].head(10)
)

print("NumPy array:")
print(temperature_array)
```

```python
print("Mean:", np.mean(temperature_array))
print("Median:", np.median(temperature_array))
print("Sum:", np.sum(temperature_array))
print("Standard deviation:", np.std(temperature_array))
print("Unique values:", np.unique(temperature_array))
```

```python
print("First value:", temperature_array[0])

print("First five values:")
print(temperature_array[0:5])
```

```python
temperature_array = np.array([30, 32, 34, 36, 38])

result = temperature_array + 2

print("Original array:")
print(temperature_array)

print("\nAfter broadcasting:")
print(result)
```

```python
df['T2M_MAX_Normalized'] = (
    (df['T2M_MAX'] - df['T2M_MAX'].min()) /
    (df['T2M_MAX'].max() - df['T2M_MAX'].min())
)

display(
    df[['T2M_MAX', 'T2M_MAX_Normalized']].head()
)
```

```python
city_data = df[['city', 'lat', 'lon']].drop_duplicates(
    subset=['city']
).reset_index(drop=True)

display(city_data.head())
```

```python
city_info = city_data[['city', 'lat', 'lon']]

merged_df = pd.merge(
    df,
    city_info,
    on='city',
    how='left',
    suffixes=('', '_info')
)

display(
    merged_df[['city', 'lat', 'lon', 'T2M_MAX']].head()
)
```

```python
df1 = df[['city', 'T2M_MAX']].head(5).set_index('city')

df2 = df[['city', 'RH2M']].head(5).set_index('city')

joined_df = df1.join(
    df2,
    lsuffix='_temperature',
    rsuffix='_humidity'
)

display(joined_df)
```

```python
df_part1 = df.head(5)

df_part2 = df.tail(5)

concatenated_df = pd.concat(
    [df_part1, df_part2],
    ignore_index=True
)

display(concatenated_df)
```
