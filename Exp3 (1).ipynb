```python
data = df['T2M_MAX'].dropna().tolist()
def calculate_mean(data):
    total = 0

    for value in data:
        total += value

    mean = total / len(data)
    return mean

def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)

    if n % 2 == 0:
        middle1 = sorted_data[n // 2 - 1]
        middle2 = sorted_data[n // 2]
        median = (middle1 + middle2) / 2
    else:
        median = sorted_data[n // 2]

    return median

def calculate_mode(data):
    frequency = {}

    for value in data:
        value = round(value, 1)

        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

    mode = None
    max_frequency = 0

    for value in frequency:
        if frequency[value] > max_frequency:
            max_frequency = frequency[value]
            mode = value

    return mode, max_frequency

def calculate_variance(data, mean):
    total = 0

    for value in data:
        total += (value - mean) ** 2

    variance = total / (len(data) - 1)
    return variance

def calculate_standard_deviation(variance):
    standard_deviation = variance ** 0.5
    return standard_deviation


def calculate_quartiles(data):
    sorted_data = sorted(data)
    n = len(sorted_data)

    position_q1 = (n + 1) * 0.25
    lower_index = int(position_q1) - 1
    fraction_q1 = position_q1 - int(position_q1)

    if lower_index + 1 < n:
        Q1 = sorted_data[lower_index] + fraction_q1 * (
            sorted_data[lower_index + 1] - sorted_data[lower_index]
        )
    else:
        Q1 = sorted_data[lower_index]

    position_q3 = (n + 1) * 0.75
    lower_index = int(position_q3) - 1
    fraction_q3 = position_q3 - int(position_q3)

    if lower_index + 1 < n:
        Q3 = sorted_data[lower_index] + fraction_q3 * (
            sorted_data[lower_index + 1] - sorted_data[lower_index]
        )
    else:
        Q3 = sorted_data[lower_index]

    return Q1, Q3

def calculate_iqr(Q1, Q3):
    return Q3 - Q1

mean = calculate_mean(data)

median = calculate_median(data)

mode, mode_frequency = calculate_mode(data)

variance = calculate_variance(data, mean)

standard_deviation = calculate_standard_deviation(variance)

Q1, Q3 = calculate_quartiles(data)

IQR = calculate_iqr(Q1, Q3)

print("STATISTICAL ANALYSIS OF T2M_MAX")
print("----------------------------------------")
print("Number of observations :", len(data))
print("Mean                  :", round(mean, 4), "°C")
print("Median                :", round(median, 4), "°C")
print("Mode                  :", mode, "°C")
print("Mode Frequency        :", mode_frequency)
print("Variance              :", round(variance, 4), "°C²")
print("Standard Deviation    :", round(standard_deviation, 4), "°C")
print("Q1                    :", round(Q1, 4), "°C")
print("Q3                    :", round(Q3, 4), "°C")
print("Interquartile Range   :", round(IQR, 4), "°C")
```

```python
temperature = df['T2M_MAX'].dropna()
bins = [0, 10, 15, 20, 25, 30, 35, 40, 45, 50]
labels = [
    '0-10',
    '10-15',
    '15-20',
    '20-25',
    '25-30',
    '30-35',
    '35-40',
    '40-45',
    '45-50'
]

groups = pd.cut(temperature, bins=bins, labels=labels,
                include_lowest=True, right=False)

frequency = groups.value_counts().sort_index()

midpoints = [5, 12.5, 17.5, 22.5, 27.5, 32.5, 37.5, 42.5, 47.5]

total_frequency = 0
weighted_sum = 0

for i in range(len(frequency)):
    f = frequency.iloc[i]
    m = midpoints[i]

    total_frequency += f
    weighted_sum += f * m

grouped_mean = weighted_sum / total_frequency

variance_sum = 0

for i in range(len(frequency)):
    f = frequency.iloc[i]
    m = midpoints[i]

    variance_sum += f * ((m - grouped_mean) ** 2)

grouped_variance = variance_sum / (total_frequency - 1)

grouped_std = grouped_variance ** 0.5

cumulative_frequency = []
total = 0

for f in frequency:
    total += f
    cumulative_frequency.append(total)

N = total_frequency
median_position = N / 2

median_class_index = 0

for i in range(len(cumulative_frequency)):
    if cumulative_frequency[i] >= median_position:
        median_class_index = i
        break

lower_boundaries = [0, 10, 15, 20, 25, 30, 35, 40, 45]

L = lower_boundaries[median_class_index]
f_median = frequency.iloc[median_class_index]

if median_class_index == 0:
    cf_previous = 0
else:
    cf_previous = cumulative_frequency[median_class_index - 1]

class_width = bins[median_class_index + 1] - bins[median_class_index]

grouped_median = L + (
    (median_position - cf_previous) / f_median
) * class_width


modal_class_index = frequency.argmax()

L_mode = lower_boundaries[modal_class_index]

f1 = frequency.iloc[modal_class_index]

if modal_class_index == 0:
    f0 = 0
else:
    f0 = frequency.iloc[modal_class_index - 1]

if modal_class_index == len(frequency) - 1:
    f2 = 0
else:
    f2 = frequency.iloc[modal_class_index + 1]

h = class_width

grouped_mode = L_mode + (
    (f1 - f0) / ((2 * f1) - f0 - f2)
) * h

Q1_position = N / 4
Q1_class_index = 0

for i in range(len(cumulative_frequency)):
    if cumulative_frequency[i] >= Q1_position:
        Q1_class_index = i
        break

L_Q1 = lower_boundaries[Q1_class_index]
f_Q1 = frequency.iloc[Q1_class_index]

if Q1_class_index == 0:
    cf_Q1 = 0
else:
    cf_Q1 = cumulative_frequency[Q1_class_index - 1]

Q1 = L_Q1 + (
    (Q1_position - cf_Q1) / f_Q1
) * class_width


Q3_position = 3 * N / 4
Q3_class_index = 0

for i in range(len(cumulative_frequency)):
    if cumulative_frequency[i] >= Q3_position:
        Q3_class_index = i
        break

L_Q3 = lower_boundaries[Q3_class_index]
f_Q3 = frequency.iloc[Q3_class_index]

if Q3_class_index == 0:
    cf_Q3 = 0
else:
    cf_Q3 = cumulative_frequency[Q3_class_index - 1]

Q3 = L_Q3 + (
    (Q3_position - cf_Q3) / f_Q3
) * class_width

grouped_IQR = Q3 - Q1

print("GROUPED FREQUENCY DISTRIBUTION")
print("----------------------------------------")
print("Temperature Range\tFrequency")

for i in range(len(frequency)):
    print(labels[i], "\t\t", frequency.iloc[i])


print("\nGROUPED DATA STATISTICAL ANALYSIS")
print("----------------------------------------")
print("Total Frequency       :", total_frequency)
print("Mean                  :", round(grouped_mean, 4), "°C")
print("Median                :", round(grouped_median, 4), "°C")
print("Mode                  :", round(grouped_mode, 4), "°C")
print("Variance              :", round(grouped_variance, 4), "°C²")
print("Standard Deviation    :", round(grouped_std, 4), "°C")
print("Q1                    :", round(Q1, 4), "°C")
print("Q3                    :", round(Q3, 4), "°C")
print("Interquartile Range   :", round(grouped_IQR, 4), "°C")
```
